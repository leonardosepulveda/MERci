# Select pipeline — export notebooks for one acquisition type

Copies one `before_imaging/` pipeline, plus the shared `during_imaging/` and
`after_imaging/` notebooks, into `SAMPLE_DIR/notebooks/`, next to this `MERci/`
clone. Run once per experiment, then work from `SAMPLE_DIR/notebooks/`.

## 1 — Setup

```
import os
import sys
from pathlib import Path

MERCI_DIR  = Path(os.getcwd()).parent.parent   # MERci/ (notebook lives in MERci/notebooks/before_imaging/)
SAMPLE_DIR = MERCI_DIR.parent                  # experiment root, e.g. LT048_sample_18/
sys.path.insert(0, str(MERCI_DIR / "src"))

from MERci.acquisition.pipeline_export import describe_pipelines, export_pipeline_notebooks

print(f"MERCI_DIR  : {MERCI_DIR}")
print(f"SAMPLE_DIR : {SAMPLE_DIR}")
```

## 2 — Available pipelines

One `before_imaging/` pipeline per acquisition type. `during_imaging/` and
`after_imaging/` are always exported too.

```
PIPELINE_INFO = describe_pipelines(MERCI_DIR)
for pid, info in PIPELINE_INFO.items():
    print(f"{pid}\n    source: {info.source}\n    {info.description}\n")
```

## 3 — Choose a pipeline

Set `PIPELINE` to one of the ids printed above.

```
PIPELINE = "tumor_epi"
```

## 4 — Export

Copies the notebooks into `SAMPLE_DIR/notebooks/` and points their `MERCI_DIR`
at the sibling `MERci/`. Also copies the pipeline's `pipeline.yaml` and
`round_bit_color.csv` (if it has them) there, and the notebooks read that copy,
so it can be edited per experiment. Writes `notebooks/README.md`. Fails if
`SAMPLE_DIR/notebooks/` exists; `force=True` overwrites it.

```
notebooks_dir = export_pipeline_notebooks(MERCI_DIR, SAMPLE_DIR, PIPELINE, force=False)
print(f"Exported '{PIPELINE}' to: {notebooks_dir}")
print(f"README: {notebooks_dir / 'README.md'}")
```
