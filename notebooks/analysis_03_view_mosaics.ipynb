# View mosaics

Displays all round mosaics produced by `analysis_02_round_scheduler.ipynb`.  
Re-run any cell to refresh the view as new mosaics appear.

This notebook is self-contained — it only needs the `analysis/mosaics/` folder.

```python
import os
import sys
from pathlib import Path

MERCI_DIR  = Path(os.getcwd()).parent          # MERci/ (notebook lives in MERci/notebooks/)
SAMPLE_DIR = MERCI_DIR.parent                  # experiment root
sys.path.insert(0, str(MERCI_DIR / "src"))

from MERci.visualization import display_mosaic

ANALYSIS_DIR = SAMPLE_DIR / "analysis"
MOSAICS_DIR  = ANALYSIS_DIR / "mosaics"

print(f"SAMPLE_DIR   : {SAMPLE_DIR}")
print(f"MOSAICS_DIR  : {MOSAICS_DIR}")
```

```python
# ── Display parameters ─────────────────────────────────────────────────
# Filter to specific rounds (e.g. [1, 2, 3]) or None to show all
SHOW_ROUNDS = None

# Filter to specific colors in nm (e.g. [560, 650]) or None to show all
SHOW_COLORS = None

# Width of each displayed image in pixels
DISPLAY_WIDTH = 900
```

```python
import re
from IPython.display import display, Image as IPImage

mosaic_files = sorted(MOSAICS_DIR.glob("*.png")) if MOSAICS_DIR.exists() else []

if not mosaic_files:
    print("No mosaics found yet. Run analysis_02_round_scheduler.ipynb first and wait for at least one round to complete.")
else:
    for p in mosaic_files:
        # Parse round and color from filename
        # Patterns: round_001_560nm_mosaic.png  OR  round_001_mosaic.png (legacy)
        m_color = re.match(r"round_(\d+)_(\d+)nm_mosaic\.png", p.name)
        m_plain = re.match(r"round_(\d+)_mosaic\.png", p.name)

        if m_color:
            rid   = int(m_color.group(1))
            color = int(m_color.group(2))
        elif m_plain:
            rid   = int(m_plain.group(1))
            color = None
        else:
            rid, color = 0, None

        if SHOW_ROUNDS is not None and rid not in SHOW_ROUNDS:
            continue
        if SHOW_COLORS is not None and color not in SHOW_COLORS:
            continue

        color_label = f"  {color} nm" if color is not None else ""
        print(f"Round {rid:03d}{color_label}  —  {p.name}")
        display(IPImage(str(p), width=DISPLAY_WIDTH))
```
