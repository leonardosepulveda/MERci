# Select pipeline — export notebooks for one acquisition type

Picks one `before_imaging/` pipeline and copies it, plus the shared `during_imaging/`/`after_imaging/` notebooks, into a standalone `SAMPLE_DIR/notebooks/` folder that sits *alongside* this `MERci/` clone. Run this once per experiment, then do all further work from `SAMPLE_DIR/notebooks/` — it makes which pipeline was used obvious from the folder itself, and this `MERci/` clone stays untouched.

## 1 — Setup

```
import os
import sys
from pathlib import Path

MERCI_DIR  = Path(os.getcwd()).parent.parent   # MERci/ (notebook lives in MERci/notebooks/setup/)
SAMPLE_DIR = MERCI_DIR.parent                  # experiment root, e.g. LT048_sample_18/
sys.path.insert(0, str(MERCI_DIR / "src"))

from MERci.acquisition.pipeline_export import describe_pipelines, export_pipeline_notebooks

print(f"MERCI_DIR  : {MERCI_DIR}")
print(f"SAMPLE_DIR : {SAMPLE_DIR}")
```

## 2 — Available pipelines

One `before_imaging/` variant per acquisition type. `during_imaging/` and `after_imaging/` are shared across all of them and always get exported alongside whichever one you pick below.

```
PIPELINE_INFO = describe_pipelines(MERCI_DIR)
for pid, info in PIPELINE_INFO.items():
    print(f"{pid}\n    source: {info.source}\n    {info.description}\n")
```

## 3 — Choose a pipeline

Set `PIPELINE` to one of the ids printed above.

```
PIPELINE = "reference"
```

## 4 — Export

Copies `PIPELINE`'s `before_imaging/` notebooks (flattened) plus `during_imaging/`/`after_imaging/` into `SAMPLE_DIR/notebooks/`, fixes each copy's `MERCI_DIR` line for the new sibling-`MERci` layout, and writes `notebooks/README.md`. Fails if `SAMPLE_DIR/notebooks/` already exists — pass `force=True` to overwrite a previous export.

```
notebooks_dir = export_pipeline_notebooks(MERCI_DIR, SAMPLE_DIR, PIPELINE, force=False)
print(f"Exported '{PIPELINE}' to: {notebooks_dir}")
print(f"README: {notebooks_dir / 'README.md'}")
```
